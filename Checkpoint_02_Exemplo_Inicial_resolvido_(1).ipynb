# Exemplo do uso de scipy para integrais definidas

Considere a função

$$
f(x)=exp(x)+1.
$$

Utilizando as funções `quad` e `trapezoid` da biblioteca `scipy.integrate`, calcule:

1. A integral de $f(x)$ no intervalo $[0,5]$: $\int_0^5 f(x)\,dx$

2. A integral de $f(x)$ no intervalo $[0,2]$: $\int_0^2 f(x)\,dx$

3. A diferença entre os dois valores obtidos.

4. A integral de $f(x)$ diretamente no intervalo $[2,5]$: $\int_2^5 f(x)\,dx$


5. Compare os resultados dos itens 3 e 4.

## Setup

```python
import numpy as np
import scipy.integrate
```

https://docs.scipy.org/doc/scipy/reference/integrate.html

## quad

```python
def f(x):
    return np.exp(x) + 1

# Integral de f(x) em [0, 5]
I_0_5, erro_0_5 = scipy.integrate.quad(f, 0, 5)

# Integral de f(x) em [0, 2]
I_0_2, erro_0_2 = scipy.integrate.quad(f, 0, 2)
diferenca = I_0_5 - I_0_2

# Integral de f(x) diretamente em [2, 5]
I_2_5, erro_2_5 = scipy.integrate.quad(f, 2, 5)

print("== quad ==")
print(f"Valor da integral em [0, 5]: {I_0_5:.6f}")
print(f"Valor da integral em [0, 2]: {I_0_2:.6f}")
print(f"A Diferenca (item1 - item2): {diferenca:.6f}")
print(f"Valor da integral em [2, 5]: {I_2_5:.6f}")
print(f"verificando se item 3 == item 4? {np.isclose(diferenca, I_2_5)}  "
      f"(diferenca = {abs(diferenca - I_2_5):.2e})")
```

## trapezoid

```python
def f(x):
    return np.exp(x) + 1

n = 100_000

# Integral de f(x) em [0, 5]
x_0_5 = np.linspace(0, 5, n)
I_0_5_t = scipy.integrate.trapezoid(f(x_0_5), x_0_5)

# Integral de f(x) em [0, 2]
x_0_2 = np.linspace(0, 2, n)
I_0_2_t = scipy.integrate.trapezoid(f(x_0_2), x_0_2)

# (item 1 - item 2)
diferenca_t = I_0_5_t - I_0_2_t

# Integral de f(x) diretamente em [2, 5]
x_2_5 = np.linspace(2, 5, n)
I_2_5_t = scipy.integrate.trapezoid(f(x_2_5), x_2_5)

print("== trapezoid ==")
print(f"Valor da integral em [0, 5]: {I_0_5_t:.6f}")
print(f"Valor da integral em [0, 2]: {I_0_2_t:.6f}")
print(f"(item1 - item2): {diferenca_t:.6f}")
print(f"Valor da integral em [2, 5]: {I_2_5_t:.6f}")
print(f"verificando se item 3 == item 4? {np.isclose(diferenca_t, I_2_5_t)}  "
      f"(diferenca = {abs(diferenca_t - I_2_5_t):.2e})")
print()
print(f"verificando se quad x trapezoid batem entre si: {np.isclose(I_0_5_t, I_0_5, atol=1e-4)}")
```

## Discussão:

## Entendendo os Resultados das Integrais

A análise dos valores que calculamos para as integrais nos leva a um princípio muito importante do cálculo. Pense na integral de uma função como a **"área"** que está sob o gráfico dessa função em um determinado intervalo.

Quando calculamos a integral de `f(x)` no intervalo de `0 a 5`, estamos medindo a área total da função nesse trecho. Da mesma forma, ao calcular a integral de `0 a 2`, estamos encontrando a área apenas na primeira parte desse intervalo.

### O Ponto Chave: A Aditividade da Integral

Um conceito fundamental é que as integrais são **aditivas**. Isso significa que a área total de um intervalo maior pode ser dividida e somada a partir de subintervalos. Por exemplo, a área total de `0 a 5` é exatamente igual à soma da área de `0 a 2` com a área de `2 a 5`.

Isso tem uma implicação direta nos nossos cálculos:

*   Se você subtrair a área do intervalo `0 a 2` da área total do intervalo `0 a 5`, o que sobra é precisamente a área do intervalo `2 a 5`.

### Comparando Nossos Achados

Foi exatamente isso que observamos quando comparamos os resultados:

1.  **A diferença** entre a integral de `0 a 5` e a integral de `0 a 2` (que foi nosso **item 3**).
2.  **A integral calculada diretamente** no intervalo de `2 a 5` (que foi nosso **item 4**).

Como esperado, os resultados desses dois cálculos são **praticamente idênticos**! Isso não é uma coincidência, mas sim uma demonstração clara da **propriedade aditiva dos intervalos de integração**, uma base sólida para trabalhar com integrais definidas.
